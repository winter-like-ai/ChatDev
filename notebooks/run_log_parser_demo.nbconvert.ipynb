# ChatDev 日志解析演示

这个 Notebook 展示了如何使用刚刚实现的 `log_analyzer` 模块，将 ChatDev 的 `.log` 文件解析为结构化的 JSON 格式，并输出相关的统计信息。

```python
import os
import json
import sys

# 确保能导入 log_analyzer 模块 (如果 Notebook 位于工程根目录则直接生效)
chatdev_root = os.path.abspath('.')
if chatdev_root not in sys.path:
    sys.path.insert(0, chatdev_root)

from log_analyzer import LogParser, validate
```

```python
# 设置输入日志文件路径和输出 JSON 文件路径
log_file_path = r"dataset_mini\Basic_Network_Ping_Tool_CLI_DefaultOrganization_20260331212914\Basic_Network_Ping_Tool_CLI_DefaultOrganization_20260331212914.log"
output_json_path = log_file_path.replace(".log", ".json")

print(f"输入日志路径: {log_file_path}")
print(f"输出 JSON 路径: {output_json_path}")
```

```python
# 1. 初始化解析器
parser = LogParser()

# 2. 解析文件
# parse 方法会自动将结果存入 output_json_path
parsed_data = parser.parse(log_file_path, output_json_path)
print("解析完成！")
```

```python
# 3. 验证 JSON Schema
is_valid, errors = validate(parsed_data)

if is_valid:
    print("\u2714 Schema 验证通过！输出的 JSON 格式规范。")
else:
    print("\u2718 Schema 验证失败！错误信息：")
    for error in errors:
        print(f"  - {error}")
```

```python
# 4. 打印核心统计结果 (从解析后的数据字典中提取)
summary = parsed_data.get("summary", {})

print("========== 解析摘要 ==========")
print(f"总事件数 (Total Events)  : {summary.get('total_events')}")
print(f"总 API 调用数            : {summary.get('num_api_calls')}")
print(f"总 Token 消耗量          : {summary.get('total_tokens')}")
print(f"总开销 (Cost)            : ${summary.get('total_cost')}")
print(f"执行所涉阶段 (Phases)    : {', '.join(summary.get('phases_executed', []))}")

print("\n======== 事件类型分布 ========")
event_counts = summary.get('event_type_counts', {})
for event_type, count in sorted(event_counts.items(), key=lambda x: x[1], reverse=True):
    print(f"  {event_type:<25}: {count} 次")
```

你可以打开生成的 `output_json_path` 文件查看完整而详尽的格式化数据，这个 JSON 将可被用于接下来的分析评估等场景。
