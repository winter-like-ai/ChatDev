# ChatDev Hybrid Mode Execution

This notebook demonstrates how to execute ChatDev in **Hybrid Mode**. 
In hybrid mode, the system replays the API interactions from an existing `api_records.jsonl` up to a specific node, and then seamlessly transitions to calling the **real API** for all subsequent interactions.

```python
import os

# 1. Define the snapshot/dataset directory
DATASET_DIR = r"d:\Works\code\winter-like-ai\ChatDev\dataset_mini\HYBRID_DATA\Simple_CSV_Data_Processor_CLI_DefaultOrganization_20260414171740"
JSONL_PATH = os.path.join(DATASET_DIR, "api_records.jsonl")

# 2. ChatDev task parameters used in the snapshot
PROJECT_NAME = "Simple_CSV_Data_Processor_CLI"
ORG_NAME = "DefaultOrganization"
CONFIG = "Default"
TASK_PROMPT = "Build a CLI program that reads a CSV file, allows the user to specify a column name, and then prints all unique values from that column. The program should handle CSV files with headers."

# 3. Hybrid Mode specific settings
# Which node should we start making REAL API calls? 
# For example, if HYBRID_START_NODE = 5,
# Node 1, 2, 3, 4 will be replayed entirely.
# Node 5 and onwards will use the real OpenAI API.
HYBRID_START_NODE = 5
```

### Execute `run.py` to start the Hybrid process
We use Python's `subprocess` to launch the `run.py` script and stream the output back to the cell.

```python
import sys
import subprocess

cmd_list = [
    "python", "run.py",
    "--config", CONFIG,
    "--org", ORG_NAME,
    "--name", PROJECT_NAME,
    "--task", TASK_PROMPT,
    "--hybrid", JSONL_PATH,
    "--hybrid-node", str(HYBRID_START_NODE)
]

# Print the command being executed
print("Executing:\n" + " ".join(f'\"{c}\"' if ' ' in c else c for c in cmd_list) + "\n")
print("="*60)

process = subprocess.Popen(
    cmd_list, 
    cwd=r"d:\Works\code\winter-like-ai\ChatDev",  # Make sure we run in the correct dir
    stdout=subprocess.PIPE, 
    stderr=subprocess.STDOUT, 
    text=True, 
    bufsize=1, 
    universal_newlines=True
)

for line in process.stdout:
    sys.stdout.write(line)
    sys.stdout.flush()
    
process.wait()
print(f"\nProcess exited with return code: {process.returncode}")
```
